```python
import pandas as pd
```

```python
import pathlib
pathlib.Path('.').absolute()
```

```python
embeddings_path = r'..\EvolveTest\kelsic_esm2_t36_3B_UR50D.csv'
```

```python
df = pd.read_csv(embeddings_path)
```

```python
import json
df['embedding'] = df[df.columns[1:]].apply(lambda r: json.dumps(list(r)), axis=1)
```

```python
df.rename(columns={'Unnamed: 0': 'seq_id'})[['seq_id', 'embedding']]\
    .to_csv(embeddings_path, index=False)
```
