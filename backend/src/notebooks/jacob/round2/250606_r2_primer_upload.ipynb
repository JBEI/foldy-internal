```python
import pandas as pd
# ORDER_BATCH = 'folde_r2_oligo_plate'
# idt_df = pd.read_excel('notebooks/jacob/round2/FolDE Round 2 oligos
ORDER_BATCH = 'folde_r2_oligos_dbat_oplr_random'
idt_df = pd.read_excel('notebooks/jacob/round2/folde_r2_GAH_Ap_OplR_random_R2_spec_sheet_raw.xlsx')
```

```python
def get_new_well_position(order_batch, old_plate, old_well):
    column_delta, row_delta = {
        ('folde_r2_oligo_plate', 'order table 1'): (0, 0),
        ('folde_r2_oligo_plate', 'order table 2'): (1, 0),
        ('folde_r2_oligo_plate', 'order table 3'): (0, 1),
        ('folde_r2_oligos_dbat_oplr_random', 'order table 1'): (0, 0),
    }[(order_batch, old_plate)]
        
    # Split into row and column
    row_96 = old_well[0].upper()
    try:
        col_96 = int(old_well[1:])
    except ValueError:
        raise ValueError("Invalid column number")
    
    # Validate input ranges
    if not ('A' <= row_96 <= 'H'):
        raise ValueError("Row must be between A and H")
    if not (1 <= col_96 <= 12):
        raise ValueError("Column must be between 1 and 12")
    
    # Convert row (every row in 96 maps to every other row in 384)
    row_384_num = (ord(row_96) - ord('A')) * 2
    row_384 = chr(ord('A') + row_384_num + row_delta)
    
    # Convert column (every column in 96 maps to every other column in 384)
    col_384 = (col_96 - 1) * 2 + 1 + column_delta
    
    return f"{row_384}{col_384}"

out_df = idt_df[['Sequence Name']].copy() 
out_df['Sequence'] = idt_df['Sequence'].str.replace(' ', '')
out_df['Well Position'] = idt_df.apply(lambda r: get_new_well_position(ORDER_BATCH, r['Plate Name'], r['Well Position']), axis=1)
out_df['Plate Name'] = ORDER_BATCH
out_df['Molarity nM'] = idt_df['nmoles'] * 10
out_df['ug'] = idt_df['µg']
out_df['Final Volume uL'] = 50.0
out_df
```

```python
out_df.to_csv('notebooks/jacob/round2/250606_r2_dbat_oplr_random_oligos_processed.csv')
```
