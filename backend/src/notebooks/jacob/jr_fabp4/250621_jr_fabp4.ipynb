```python
import pandas as pd
from pathlib import Path

from app.helpers.sequence_util import seq_id_to_seq

def write_bulk_input_to_file(seq_ids, file_name, ligand_name, directory='notebooks/jacob/jr_fabp4/'):
    inp_dicts = []
    
    if ligand_name == 'PFOA':
        ligand_smiles = 'C(=O)(C(C(C(C(C(C(C(F)(F)F)(F)F)(F)F)(F)F)(F)F)(F)F)(F)F)O'
    elif ligand_name == 'PFBS':
        ligand_smiles = 'C(C(C(F)(F)S(=O)(=O)O)(F)F)(C(F)(F)F)(F)F'
    else:
        raise ValueError(f'Invalid ligand: {ligand_name}')
    
    for seq_id in seq_ids:
        inp_dicts.append({
            'fold_name': f'FABP4_HUMAN-{seq_id} {ligand_name}',
            'chain_1_id': 'A',
            'chain_1_type': 'protein',
            'chain_1_sequence': seq_id_to_seq(wt_aa_seq, seq_id),
            'chain_2_id': ligand_name,
            'chain_2_type': 'ligand',
            'chain_2_smiles': ligand_smiles,
            'affinity_chain_id': ligand_name
        })
    inp_df = pd.DataFrame(inp_dicts)
    inp_df.to_csv(Path(directory) / file_name, index=False)
```
